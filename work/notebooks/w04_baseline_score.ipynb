```python
import pandas as pd

df = pd.read_csv("flyrank-ml/data/raw/content_refresh_anonymized.csv")

print(df.head())
print(df.columns.tolist())
# Refresh page if last updated>90 days ago, has low clicks but high impressions.
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
high_impressions= df["impressions_90d"].median()
lower_ctr=df["ctr"].median()
print("High impressions and low CTR = ", high_impressions," and ", lower_ctr)

a = (df["days_since_last_update"] > 90).astype(int)
b = (df["impressions_90d"] >= high_impressions).astype(int)
c = (df["ctr"] < lower_ctr).astype(int)
df["score"]=a+b+c
print(df[["days_since_last_update","impressions_90d","ctr","score"]].head())

df=df.sort_values("score", ascending=False)
# df.to_csv("work/outputs/baseline_action_score.csv", index=False)
```
